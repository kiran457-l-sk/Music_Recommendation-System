```python
df['dance_tag']=pd.cut(df['danceability'],
                        bins=[0.0,0.4,0.7,1.0],
                        labels=['dance_low','dance_mid','dance_high'],
                        include_lowest=True
                    )
```

```python
df['acoustic_tag'] = pd.cut(
    df['acousticness'],
    bins=[0.0, 0.3, 0.7, 1.0],
    labels=['acoustic_low', 'acoustic_mid', 'acoustic_high'],
    include_lowest=True
)
```

```python
df['tempo_tag'] = pd.cut(
    df['tempo'],
    bins=[0, 95, 120, 300],
    labels=['tempo_slow', 'tempo_mid', 'tempo_fast'],
    include_lowest=True
)

```

```python
df['mode_tag'] = df['mode'].map({1: 'mode_major', 0: 'mode_minor'})
```

```python
df['energy_tag'] = pd.cut(df['energy'], bins=[0.0, 0.4, 0.7, 1.0], labels=['energy_low', 'energy_mid', 'energy_high'], include_lowest=True)
df['valence_tag'] = pd.cut(df['valence'], bins=[0.0, 0.35, 0.65, 1.0], labels=['valence_sad', 'valence_neutral', 'valence_happy'], include_lowest=True)
```

```python
df['loudness_tag'] = pd.qcut(df['loudness'], q=3, labels=['loudness_quiet', 'loudness_mid', 'loudness_loud'], duplicates='drop')
```

```python
key_map = {
    -1: 'key_unknown', 0: 'key_C', 1: 'key_Csharp', 2: 'key_D', 3: 'key_Dsharp',
    4: 'key_E', 5: 'key_F', 6: 'key_Fsharp', 7: 'key_G', 8: 'key_Gsharp',
    9: 'key_A', 10: 'key_Asharp', 11: 'key_B'
}
```

```python
df['instr_tag'] = pd.cut(df['instrumentalness'], bins=[0.0, 0.5, 1.0], labels=['vocal_track', 'instrumental_track'], include_lowest=True)
df['live_tag'] = pd.cut(df['liveness'], bins=[0.0, 0.8, 1.0], labels=['studio_track', 'live_track'], include_lowest=True)
```

```python
df['key_tag'] = df['key'].map(key_map).fillna('key_unknown')
```

```python
df_unique = df.drop_duplicates(subset=['track_name', 'track_artist']).reset_index(drop=True)
```

```python
new_df=df_unique[['track_name','track_artist','playlist_name','playlist_genre','playlist_subgenre','dance_tag','acoustic_tag','tempo_tag','mode_tag','energy_tag','valence_tag','loudness_tag','instr_tag','live_tag','key_tag']]
```

```python
df_final = new_df.drop_duplicates(subset=['track_name', 'track_artist']).reset_index(drop=True)
```

```python
df_final['track_name'] = (
    df_final['track_name'].str.split(r'\s-\s|\(|\\|\.').str[0].str.strip() )
```

```python
df_clean = df_final.astype(str).replace(r'\s+', '', regex=True).apply(lambda x: x.str.lower())
```

```python
import re
def remove_emojis(text):
    cleaned = re.sub(r'[^\x00-\x7F]+', '', str(text))
    return re.sub(r'\s+', ' ', cleaned).strip()
```

```python
df_clean['track_name']=df_clean['track_name'].apply(remove_emojis)
df_clean['track_artist']=df_clean['track_artist'].apply(remove_emojis)
df_clean['playlist_name']=df_clean['playlist_name'].apply(remove_emojis)
df_clean['playlist_genre']=df_clean['playlist_genre'].apply(remove_emojis)
df_clean['playlist_subgenre']=df_clean['playlist_subgenre'].apply(remove_emojis)
```

```python
df_clean.sample(3)
```

```python
df_clean.columns
```

```python
cols_to_combine = [
    'track_artist', 'playlist_name', 'playlist_genre', 'playlist_subgenre',
    'dance_tag', 'acoustic_tag', 'tempo_tag', 'mode_tag',
    'energy_tag', 'valence_tag', 'loudness_tag', 'instr_tag', 'live_tag'
]
df_clean['bow_text'] = df_clean[cols_to_combine].astype(str).agg(' '.join, axis=1)
```

```python
df_root=df_clean[['track_name','bow_text']]
```

```python
df_root.sample(3)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer(max_features=5000,stop_words='english')
vectors=cv.fit_transform(df_root['bow_text'])
```

```python
vectors[0]
```

```python
cv.get_feature_names_out()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
df_root
```

```python
def recommend(music):
    music_index=df_root[df_root['track_name']==music].index[0]
    distances=cosine_similarity(vectors[music_index],vectors).flatten()
    music_list=sorted(list(enumerate(distances)),reverse=True,key=lambda x:x[1])[1:6]
    for i in music_list:
        print(df_root.iloc[i[0]].track_name)
```

```python
recommend('allthetime')
```
